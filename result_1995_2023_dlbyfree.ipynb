```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from typing import List, Dict, Tuple, Callable
import scipy as sp
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns
import itertools
import io
import os
from google.colab import files
```

```python
# ファイルアップロード
result_uploaded = files.upload()

# アップロードしたファイルを読み込む
result_path = list(result_uploaded.keys())[0]  # アップロードしたファイル名を取得
result = pd.read_excel(result_path, header=None)

# 行と列を交換（転置）
transposed_result = result.T

# 保存先パスを指定
save_path = "/Users/haruk/iCloudDrive/Documents/Research/99_Sotsuron/sotsuron/transposed_result_1995_2023_dlbyfree.xlsx"
os.makedirs(os.path.dirname(save_path), exist_ok=True)  # ディレクトリが存在しない場合は作成
transposed_result.to_excel(save_path, index=False, header=False)

print(f"行と列を交換したデータを {save_path} に保存しました。")
```

```python
# 列名が値に含まれている場合の処理
# データフレームの最初の行を列名として設定
transposed_result.columns = transposed_result.iloc[0]  # 最初の行を列名に設定
transposed_result = transposed_result[1:]  # 最初の行を削除
transposed_result.head()
```

```python
# 必要なデータを抽出
years = transposed_result['year']  # 'year'列を横軸として使用

# 'year' 以外のすべての列で折れ線グラフを作成
for column in transposed_result.columns:
    if column != 'year':  # 'year' 列はスキップ
        flow_values = transposed_result[column].astype(float)  # 対応する列を縦軸として使用

        # 折れ線グラフを作成
        plt.figure(figsize=(10, 6))
        plt.plot(years, flow_values, marker='o', linestyle='-', label=column)
        plt.xlabel('Year')
        plt.ylabel('Value')
        plt.title(f"{column} over Years")
        plt.grid(True)
        plt.legend()
        plt.tight_layout()
        plt.show()
```
