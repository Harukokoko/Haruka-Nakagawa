```python
import pandas as pd
import networkx as nx
import matplotlib.pyplot as plt
from typing import List, Dict, Tuple, Callable
import scipy as sp
import numpy as np
import math
import matplotlib.pyplot as plt
import seaborn as sns
import itertools
import io
import os
from google.colab import files
```

```python
# ファイルアップロード
result_uploaded = files.upload()

# アップロードしたファイルを読み込む
result_path = list(result_uploaded.keys())[0]  # アップロードしたファイル名を取得
result = pd.read_excel(result_path, header=None)
```

```python
result.columns = result.iloc[0]  # 最初の行を列名に設定
result = result[1:]  # 最初の行を削除
result.head()
```

```python
# HScodeのユニークな値を取得
hscodes = result['HScode'].unique()

# ① yearを横軸とし、各指標をプロット
columns_1 = [
    "The calculated flow into 'usa' from 'china'",
    "The direct export ratio from 'china' to 'usa'",
    "The_ratio_of_di",
    "The_ratio_of_indi",
    "Indi_to_di"
]

for column in columns_1:
    plt.figure(figsize=(10, 6))
    for hscode in hscodes:
        filtered_data = result[result['HScode'] == hscode]
        plt.plot(filtered_data['year'], filtered_data[column], label=f'HScode: {hscode}')

    plt.title(f'Trends by Year for {column}')
    plt.xlabel('Year')
    plt.ylabel(column)
    plt.legend()
    plt.grid(True)
    plt.show()

# ② China's total export volumeとThe estimated final flow into 'usa' from 'china'をプロット
columns_2 = [
    "China's total export volume",
    "The estimated final flow into 'usa' from 'china'"
]

for hscode in hscodes:
    filtered_data = result[result['HScode'] == hscode]

    plt.figure(figsize=(10, 6))
    for column in columns_2:
        plt.plot(filtered_data['year'], filtered_data[column], label=column)

    plt.title(f'HScode: {hscode} - Export Volume and Final Flow Trends')
    plt.xlabel('Year')
    plt.ylabel('Values')
    plt.legend()
    plt.grid(True)
    plt.show()
```
